# Least-Cost Modelling of Gravettian Palaeocoastal Accessibility
<br>

## Project Overview

This project investigates the potential accessibility of the Mediterranean palaeocoast from **Cova de la Barriada**, a Gravettian archaeological site located in the Serra Gelada (Benidorm, Spain).

Archaeological evidence from the site, discussed by Fernández-López de Pablo et al. (2014), includes abundant remains of the terrestrial land snail *Iberus alonensis*, interpreted as evidence of dietary diversification during the Early Upper Palaeolithic. During the period of occupation, sea level was approximately **100 m below its present position**, placing the Mediterranean coastline considerably farther from the cave. Reconstructing the location and accessibility of this palaeocoast therefore provides a spatial context for assessing potential access to coastal resources.

This project reconstructs the Gravettian palaeolandscape using **official topographic and bathymetric data**, defining the palaeocoast approximately along the **−100 m bathymetric contour**. The reconstructed coastline is subsequently divided into spatial sectors to evaluate accessibility to different portions of the palaeocoast.

A terrain-based least-cost analysis is used to model the relative difficulty of movement from the cave towards each coastal sector. The analysis identifies the **most accessible coastal sector**, together with its corresponding least-cost access point and route from the cave. The model is based exclusively on the physical characteristics of the reconstructed terrain and does not incorporate variables such as vegetation or land use, for which reliable Gravettian spatial data are unavailable.

The analysis is implemented as a **reproducible and reusable ArcGIS Pro Notebook workflow**. Processing stages are explicitly defined and parameterised, original source data are preserved unchanged, and intermediate and derived datasets are managed through a dedicated staging geodatabase. Project-relative paths and standardised dataset schemas allow the workflow to be rerun from the original inputs while maintaining a consistent and traceable processing sequence.

**Research question:**
*Which sector of the Gravettian palaeocoast was most accessible from Cova de la Barriada, where was its least-cost access point, and what route provided the lowest-cost connection from the cave?*

# Data Preparation

>This phase covers the preparation and integration of the acquired spatial datasets prior to surface generation and subsequent spatial analysis. The datasets are spatially standardised, subset to the area required for interpolation, and selectively filtered to retain the elevation and bathymetric features relevant to the study. The prepared topographic and bathymetric datasets are then integrated into a common spatial framework, providing the inputs required for the generation of a continuous terrestrial–marine elevation model.

### Project Extent Definition

The project extent was manually delineated in ArcGIS Pro using the datasets acquired during the Data Acquisition stage as the primary spatial reference, particularly the bathymetric contours and surrounding topography. The −100 m depth contour was used as a guide for defining the offshore limit. The resulting polygon was stored as `project_extent.shp` under `/DATA/REFERENCE` and defines the broader spatial framework for subsequent data preparation and surface generation.

### Data Standardization and Integration

This stage prepares and integrates the acquired datasets for subsequent spatial analysis. Input datasets are standardised within a common coordinate reference system (EPSG:25830), assigned consistent project-specific names, spatially restricted to the predefined project extent, and converted to 2D geometry to reduce data volume and ensure compatibility between datasets. Where required, source-specific attribute criteria are applied to retain only the features relevant to the study. For the topographic dataset, contour features classified as standard (`TIPO_0201 = '01'`) are retained, while bathymetric contours (`TIPO_0201 = '04'`) and others are excluded, following the [BTN Product Specifications](https://www.ign.es/resources/docs/IGNCnig/BTN/ESPBTN.pdf). The attribute schema of the topographic and bathymetric datasets is also standardised, retaining only the elevation attribute and harmonising its name as `ELEVATION`. The resulting contour datasets retain their respective native intervals of 10 m for topography and 5 m for bathymetry, providing the elevation inputs for subsequent terrain-surface interpolation.

The resulting datasets are stored in a dedicated staging geodatabase while preserving the original source data unchanged. The standardised topographic and bathymetric contours are then merged into a single `extent_contours` dataset for subsequent analysis.

In addition to the externally sourced datasets, the workflow relies on a set of manually prepared reference datasets stored in the project `REFERENCE` directory. These datasets define fixed spatial inputs and analytical parameters, including the cave location, project extent, terrain-processing boundaries, reconstructed palaeocoast, coastal sectors, and predefined access points. They are retained as part of the project to ensure that the notebook can be reproduced without requiring these manually prepared inputs to be recreated during each execution.

```python
from pathlib import Path

print("Starting data standardisation and integration...")

# Access the current ArcGIS Pro project and define the project directory
aprx = arcpy.mp.ArcGISProject("CURRENT")
PROJECT_DIR = Path(aprx.homeFolder)


# Define and create the staging geodatabase
staged_gdb = PROJECT_DIR / "STAGING" / "STAGING.gdb"

if not staged_gdb.exists():
    staged_gdb.parent.mkdir(parents=True, exist_ok=True)
    arcpy.management.CreateFileGDB(str(staged_gdb.parent), staged_gdb.name)

new_database = {'databasePath' : str(staged_gdb), 'isDefaultDatabase': False}

if new_database not in aprx.databases:
    databases = aprx.databases + [new_database]    
    aprx.updateDatabases(databases)


# Define a reusable geoprocessing environment for the project workflow.
# The function encapsulates the common environment settings applied to
# derived datasets, including the output CRS, workspace, geometry dimensions,
# overwrite behaviour, and map output handling. Additional environment
# parameters, such as processing extent, can be applied temporarily through
# a nested EnvManager when required.

def set_project_environment():
    return arcpy.EnvManager(addOutputsToMap=False,
                            overwriteOutput=True,                            
                            outputCoordinateSystem=arcpy.SpatialReference(25830),
                            workspace=str(staged_gdb),
                            scratchWorkspace=str(staged_gdb),
                            outputZFlag="Disabled",
                            outputMFlag="Disabled")

    
# Define and validate the project input datasets

# External source data
topo_raw = PROJECT_DIR / "DATA" / "RAW" / "BTN0201L_CUR_NIV.shp"
bath_raw = PROJECT_DIR / "DATA" / "RAW" / "Bati_plata.shp"

# Predefined project reference datasets
cave_reference = PROJECT_DIR / "DATA" / "REFERENCE" / "cave_location.shp"
extent_reference = PROJECT_DIR / "DATA" / "REFERENCE" / "project_extent.shp"
dtm_boundary = PROJECT_DIR / "DATA" / "REFERENCE" / "dtm_boundary.shp"
dtm_coast = PROJECT_DIR / "DATA" / "REFERENCE" / "dtm_coast.shp"
dtm_cliff = PROJECT_DIR / "DATA" / "REFERENCE" / "dtm_cliff.shp"
palaeo_coast = PROJECT_DIR / "DATA" / "REFERENCE" / "palaeo_coast.shp"
palaeo_coast_sectors = PROJECT_DIR / "DATA" / "REFERENCE" / "palaeo_coast_sectors.shp"
access_point_outward = PROJECT_DIR / "DATA" / "REFERENCE" / "access_point_outward.shp"
access_point_return  = PROJECT_DIR / "DATA" / "REFERENCE" / "access_point_return.shp"

# Validate that all required input datasets are available
required_datasets = {topo_raw : topo_raw.exists(),
                     bath_raw : bath_raw.exists(), 
                     cave_reference : cave_reference.exists(),
                     extent_reference : extent_reference.exists(),
                     dtm_boundary : dtm_boundary.exists(),
                     dtm_coast : dtm_coast.exists(),
                     dtm_cliff : dtm_cliff.exists(),
                     palaeo_coast : palaeo_coast.exists(),
                     palaeo_coast_sectors : palaeo_coast_sectors.exists(),
                     access_point_outward : access_point_outward.exists(),
                     access_point_return : access_point_return.exists()}
                    

if not all (required_datasets.values()):
    missing = [path for path, exists in required_datasets.items() if not exists]
    raise FileNotFoundError("Missing required datasets:\n" + "\n".join(f"  - {path}" for path in missing))


# Define standardised names for the staged datasets
input_datasets = list(required_datasets.keys())
staged_names = ["topo_contours", "bath_contours", "cave_location", "project_extent", 
                "dtm_boundary", "dtm_coast", "dtm_cliff", "palaeo_coast",
               "palaeo_coast_sectors", "access_point_outward", "access_point_return"]


# Run the standardisation workflow using the reusable project environment.
# Apply the project extent as an additional process-specific spatial constraint
# to restrict outputs to the predefined project extent.

with set_project_environment():
    with arcpy.EnvManager(extent=str(extent_reference)):
        
   
        for dataset, name in zip(input_datasets, staged_names):
            
            # Topography: retain only standard contour lines (TIPO_0201 = '01'),
            # exclude bathymetric contours, and standardise the elevation field name.
            if dataset == topo_raw:
                arcpy.conversion.ExportFeatures(str(dataset), name, where_clause="TIPO_0201 = '01'")
                arcpy.management.DeleteField(name, "COTA_0201", "KEEP_FIELDS")
                arcpy.management.AlterField(name, "COTA_0201", "ELEVATION", "ELEVATION")
    
            # Bathymetry: retain only the contour field and standardise its name
            # to ELEVATION to match the topographic dataset schema.
            elif dataset == bath_raw:
                arcpy.management.CopyFeatures(str(dataset), name)
                arcpy.management.DeleteField(name, "Contour", "KEEP_FIELDS")
                arcpy.management.AlterField(name, "Contour", "ELEVATION", "ELEVATION")
    
            # Reference datasets: copy without attribute schema modification.
            else:
                arcpy.management.CopyFeatures(str(dataset), name)
    
        # Integrate the standardised topographic and bathymetric contours.
        arcpy.management.Merge(["topo_contours", "bath_contours"], "extent_contours")
    
print("Data standardisation and integration completed.")
    
```

# Data Processing

>This phase transforms the standardised spatial inputs into the continuous terrain and travel-time surfaces required for subsequent least-cost modelling. The processing workflow generates a digital terrain model representing the reconstructed terrestrial–marine landscape and derives accumulated travel-time surfaces that incorporate terrain slope and coastal barriers. These outputs provide the spatial and cost surfaces required to model movement between the archaeological site and the reconstructed palaeocoast.



### Digital Terrain Model

The digital terrain model (DTM) is generated using ArcGIS Pro's `TopoToRaster` tool, which interpolates a hydrologically consistent elevation surface from elevation-controlling feature data. The interpolation is constrained by four supporting datasets, with the standardised contour dataset generated during the Data Standardization and Integration stage and the remaining tool inputs maintained as reusable reference datasets in `/DATA/REFERENCE`.

* **`extent_contours`** provides the primary elevation information for the interpolation. It combines the standardised topographic and bathymetric contours produced during Data Standardization and Integration, with elevation values stored in the harmonised `ELEVATION` field. The dataset is maintained in the staging geodatabase and used as the elevation-controlling input for surface generation.

* **`dtm_boundary`** defines the spatial extent of the resulting DTM and excludes the northwestern mountainous sector, which would increase computational cost without contributing relevant information to the subsequent mobility analysis. The boundary was created by manually editing a copy of `project_extent` and refining its geometry with additional vertices. It is smaller than `project_extent`, allowing contours outside the DTM boundary to contribute supporting elevation information to the interpolation.

* **`dtm_coast`** defines the palaeocoastal constraint corresponding to the −100 m bathymetric contour. It represents the terrestrial area landward of the reconstructed palaeocoast and is used by `TopoToRaster` to constrain the interpolated surface. The underlying `palaeo_coast` line is maintained separately in `/DATA/REFERENCE` and represents the reconstructed coastline used subsequently as the destination feature in the `Optimal Path As Line` analysis.

* **`dtm_cliff`** represents the identified coastal cliff features and constrains the interpolation across abrupt elevation discontinuities. The polyline is oriented with the lower-elevation side to the left and the higher-elevation side to the right when following the direction of the line.

The supporting spatial constraints (`dtm_boundary`, `dtm_coast`, `palaeo_coast`, and `dtm_cliff`) are maintained directly in `/DATA/REFERENCE` as reusable project inputs. 

The output DTM is generated at a 25 m cell resolution. The relatively gradual bathymetric gradient across the offshore portion of the project area supports the use of this resolution, as a finer cell size would not provide an increase in meaningful elevation information.

***TopoToRaster required approximately 20 minutes to generate the final 25 m DTM. For reproducibility and accessibility, the default CELL_SIZE is therefore set to 150 m; set it to 25 m to reproduce the final analysis.***


The `tolerance_1` and `tolerance_2` parameters are set to 5 m and 100 m, following the recommended values for contour-based inputs.


```python
CELL_SIZE = 150 # Configurable output DTM cell size in metres

print("Starting DTM generation. This may take several minutes...")

# Configure the geoprocessing environment for the staging geodatabase
with set_project_environment():
                

    # Build the TopoToRaster input specification: elevation contours, DTM boundary, coastal area and cliffs
    topo_inputs = f"extent_contours ELEVATION CONTOUR; dtm_boundary # BOUNDARY; dtm_coast # COAST; dtm_cliff # CLIFF"
    
    # Generate the DTM using the contour-specific tolerance settings
    arcpy.ddd.TopoToRaster(in_topo_features=topo_inputs,
                           out_surface_raster="dtm_raw",
                           cell_size=CELL_SIZE,
                           data_type="CONTOUR",
                           tolerance_1=5,
                           tolerance_2=100)
    
    # Copy the generated raster to create the analysis-ready DTM
    arcpy.management.CopyRaster("dtm_raw", "dtm")

print("DTM generation completed.") 
```

## Distance Accumulation

Two accumulated travel-time rasters are generated using ArcGIS Pro's `Distance Accumulation` tool to model terrestrial mobility between the archaeological site and the reconstructed palaeocoast. The analysis uses the `dtm` raster both as the surface over which distance is measured and as the vertical input for modelling the effect of terrain slope on walking time. The `dtm_cliff` dataset is supplied as a barrier to prevent movement across the identified coastal cliffs of Serra Gelada. The `Hiking Time` vertical factor, based on Tobler's hiking function, is applied to model the effect of vertical movement on walking speed and travel time. The function is defined for vertical movement angles between −70° and +70°, with the default cutoff values retained. The analysis uses a planar distance method, appropriate for the spatial extent of the study area.

The resulting accumulated travel-time rasters represent the minimum estimated travel time, expressed in hours, between each cell and the respective source configuration. By incorporating the effect of terrain slope directly through the vertical factor, the analysis avoids the need for a separately reclassified slope raster.

Two source configurations are used:

- **`cave_location` as source** — calculates the minimum estimated travel time from the archaeological site to each cell in the study area. The resulting `dist_accum_cave` raster is used in subsequent modelling of the outward journey towards the reconstructed palaeocoast, including the identification of the optimal access point and least-time route.

- **`palaeo_coast` as source** — calculates the minimum estimated travel time from each cell in the study area to the reconstructed −100 m palaeocoast. As the complete palaeocoast is used as a continuous source, the resulting `dist_accum_coast` raster represents accessibility to the reconstructed coastline and provides the travel-time component for the return journey from the palaeocoast to the archaeological site. It is also subsequently used to analyse accessibility patterns and sectorisation along the reconstructed coastline.

For each source configuration, a corresponding back-direction raster is generated: `back_direction_cave` and `back_direction_coast`. These rasters record the direction of movement required from each cell to follow the minimum accumulated travel-time path towards the respective source. They are used in subsequent modelling with `Optimal Path As Line` to reconstruct least-time routes between the archaeological site and the reconstructed palaeocoast.


```python
print ("Starting Distance Accumulation...")

# Configure the geoprocessing environment for the staging geodatabase
with set_project_environment():

    
    # Calculate accumulated travel time from the archaeological site
    # using the DTM and Hiking Time vertical factor
    arcpy.sa.DistanceAccumulation(
        in_source_data="cave_location",
        in_barrier_data="dtm_cliff",
        in_surface_raster="dtm",
        in_vertical_raster="dtm",
        vertical_factor=arcpy.sa.VfHikingTime(-70, 70),
        distance_method="PLANAR",
        out_back_direction_raster="back_direction_cave"
    ).save("dist_accum_cave")

    
    # Calculate accumulated travel time from the reconstructed palaeocoast
    # to model the return journey and spatial accessibility to the coastline
    arcpy.sa.DistanceAccumulation(
        in_source_data="palaeo_coast",
        in_barrier_data="dtm_cliff",
        in_surface_raster="dtm",
        in_vertical_raster="dtm",
        vertical_factor=arcpy.sa.VfHikingTime(-70, 70),
        distance_method="PLANAR",
        out_back_direction_raster="back_direction_coast"
    ).save("dist_accum_coast")

print ("Distance Accumulation completed.")

    
```

# Data Modelling
>This phase applies the derived terrain and travel-time surfaces to model accessibility between Cova de la Barriada and the reconstructed palaeocoast. The modelling workflow evaluates least-time routes, spatial accessibility from the archaeological site, and the relative accessibility of different sectors of the palaeocoast.
>
>The resulting models provide complementary measures of accessibility, including directional least-time paths, travel-time isochrones, and coastal accessibility zones. Together, these outputs support the identification and interpretation of the most accessible routes and areas within the reconstructed landscape.

## Travel-Time Modelling

The two accumulated travel-time surfaces and their corresponding back-direction rasters are used to model terrestrial travel between the archaeological site and the reconstructed palaeocoast. `Optimal Path As Line` is applied in both directions using the `Best Single` path type: from `cave_location` to `palaeo_coast` and from `palaeo_coast` to `cave_location`. This produces two directional least-time routes and allows the travel time and optimal access point to be evaluated independently for the outward and return journeys.

The resulting routes and travel-time values are used to derive three related outputs: the least-time route in each direction, the corresponding optimal palaeocoast access point, and the estimated round-trip travel time. Because the `Hiking Time` factor accounts for the direction of vertical movement, the outward and return journeys may follow different routes and may have different travel times.

```python
print ("Starting Optimal Path As Line...")


# Configure the geoprocessing environment for the staging geodatabase
with set_project_environment():

    # Derive the least-time route from the archaeological site to the palaeocoast
    arcpy.sa.OptimalPathAsLine(in_destination_data="palaeo_coast",
                               in_distance_accumulation_raster="dist_accum_cave",
                               in_back_direction_raster="back_direction_cave",
                               out_polyline_features="optimal_path_CAVE_COAST")
    # Derive the least-time return route from the palaeocoast to the archaeological site
    arcpy.sa.OptimalPathAsLine(in_destination_data="cave_location",
                               in_distance_accumulation_raster="dist_accum_coast",
                               in_back_direction_raster="back_direction_coast",
                               out_polyline_features="optimal_path_COAST_CAVE")
                               
                               
print ("Optimal Path As Line completed.")
                           
```

The optimal access points for the outward and return journeys are separated by approximately 56 m, indicating a small directional asymmetry in the least-time solution. Despite this limited spatial difference, travel times differ by 0.24 hours (approximately 14 minutes), mainly due to the asymmetric cost of vertical movement. This difference is concentrated around the steep terrain surrounding the archaeological site. Interestingly, the return route follows a steeper section despite being the uphill journey, illustrating that Hiking Time minimises travel time rather than slope alone and that route selection depends on both slope magnitude and direction.

## Cave Accessibility Isochrones

Travel-time isochrone zones were derived from the accumulated travel-time raster generated from the cave location. One-hour travel-time intervals were used to delineate successive areas of accessibility from the cave. These isochrone zones provide a spatial representation of travel-time accessibility and facilitate the interpretation and cartographic presentation of the resulting travel-time surface.


```python
print ("Starting Isochrone Zones generation...")

# Configure the geoprocessing environment for the staging geodatabase
with set_project_environment():
    
    # Generate 1-hour travel-time isochrone zones from the cave accessibility raster
    arcpy.sa.Contour(in_raster="dist_accum_cave",
                     out_polyline_features="isochrones_zones_cave",
                     contour_interval=1,
                     contour_type="CONTOUR_POLYGON")                    
                     
print ("Isochrone Zones generation completed.")
                     
```

## Palaeocoast Accessibility Zonation

The palaeocoast was divided into two coastal sectors (A and B) based on the spatial pattern of the accumulated travel-time surface. The segmentation point was located at `0.2410419°E, 38.3708669°N`.

Distance Allocation was then used to assign each location within the study area to the coastal sector that could be reached with the lowest accumulated travel time. The resulting surface therefore identifies areas where Sector A or Sector B represents the most accessible portion of the palaeocoast. Sector A includes the coastal segment associated with Cova de la Barriada.

The analysis accounts for terrain elevation, cliff barriers, and hiking-time effects when determining relative accessibility to the two coastal sectors.

```python
print ("Starting Distance Allocation...")

with set_project_environment():

    # Calculate terrain-based accessibility allocation to palaeocoast sectors
    # using cliff barriers and hiking-time effects.
    arcpy.sa.DistanceAllocation(in_source_data="palaeo_coast_sectors",
                                source_field="SECTOR",
                                in_barrier_data="dtm_cliff",
                                in_surface_raster="dtm",
                                in_vertical_raster="dtm",
                                vertical_factor=arcpy.sa.VfHikingTime(-70, 70),
                                distance_method="PLANAR").save("allocation_coast")

    # Convert the accessibility allocation raster to polygons for cartographic use.
    arcpy.conversion.RasterToPolygon(in_raster="allocation_coast",
                                     out_polygon_features="allocation_coast_polygon",
                                     simplify="SIMPLIFY")
                                     
                                     
                                     
print ("Distance Allocation completed.")  
```

## Final Datasets

These include the reconstructed palaeo_coast, accessibility sectors, travel-time isochrones, optimal paths, access points, cave location and dtm_coast.

The contour dataset is additionally clipped to dtm_boundary to retain both topographic and bathymetric contours for the final maps.


```python
print ("Starting Datasets Export...")

# Configure the project geoprocessing environment and restrict processing
# to the emerged land area represented by dtm_coast

with set_project_environment():
    with arcpy.EnvManager(extent="dtm_coast"):
        
        # Clip the combined topographic and bathymetric contours to the DTM boundary
        # to retain only the contours relevant to the study area
        arcpy.analysis.Clip("extent_contours", "dtm_boundary", "topo_bath_contours")


        # Define the datasets to be exported and their output directory
        datasets = ["topo_bath_contours","allocation_coast_polygon",
                    "isochrones_zones_cave", "dtm_coast", "optimal_path_CAVE_COAST",
                    "optimal_path_COAST_CAVE", "access_point_outward", "access_point_return",
                    "palaeo_coast", "dtm_boundary", "cave_location"]
        
        DATASETS_PATH  = [staged_gdb / dataset_name for dataset_name in datasets]
        
        OUTPUT_DIR = PROJECT_DIR / "OUTPUT"
        
        if not OUTPUT_DIR.exists():
            OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
            
        OUTPUT_DATASETS_PATH = [str( OUTPUT_DIR / f"{dataset_name}.shp") for dataset_name in datasets]

        # Export the selected datasets as individual shapefiles
        for dataset_path, output_path in zip(DATASETS_PATH, OUTPUT_DATASETS_PATH):
            arcpy.conversion.ExportFeatures(dataset_path, output_path)
            
print ("Datasets Export completed.")

            
```

## Map Preparation

A predefined ArcGIS Pro map template is imported into the current project to provide the final cartographic structure, including layer order and symbology. The template data connections are then redirected to the project’s generated outputs, ensuring that the map references the datasets produced by the workflow. Finally, the corresponding layout template is imported and configured to reuse the existing map, preserving the predefined map frame and cartographic composition.

```python
print ("Removing existing map and layout, if any...")

# Define map and layout templates
map_template = PROJECT_DIR / "TEMPLATES" / "COVA_BARRIADA_MAP.mapx"
layout_template = PROJECT_DIR / "TEMPLATES" / "COVA_BARRIADA_LAYOUT.pagx"

if not map_template.exists() or not layout_template.exists():
    raise FileNotFoundError("Map or layout template not found.")

# Remove existing map and layout to prevent duplicates

for layout_obj in aprx.listLayouts("COVA_BARRIADA_LAYOUT"):
    aprx.deleteItem(layout_obj)
    
for map_obj in aprx.listMaps("COVA_BARRIADA_MAP"):
    aprx.deleteItem(map_obj)

print("Existing map and layout removed successfully.")



```

```python
print("Starting Map and Layout Import...")

# Import the map template
cova_map = aprx.importDocument(str(map_template))

# Import the layout template and reuse the existing map
cova_layout = aprx.importDocument(str(layout_template), reuse_existing_maps=True)

# Open map and layout views
cova_map.openView()
cova_layout.openView()

print("Map and Layout preparation completed.")
```
